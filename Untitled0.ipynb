```python
# Step 3: Train YOLO classification model
from ultralytics import YOLO
import torch

# Clear any cached memory
torch.cuda.empty_cache()

# Load pre-trained classification model (nano version - fastest for mobile)
model = YOLO('yolov8n-cls.pt')

print("="*50)
print("🚀 STARTING TRAINING")
print("="*50)
print(f"Dataset: /content/currency_detection-1")
print(f"Classes: Rs.10, Rs.20, Rs.50, Rs.100, Rs.200, Rs.500")
print(f"GPU: Tesla T4")
print(f"Images: 16,264")
print("="*50)

# Train the model
results = model.train(
    data='/content/currency_detection-1',  # Path to your dataset
    epochs=50,                    # Number of training epochs
    imgsz=224,                     # Image size
    batch=32,                       # Batch size (good for T4)
    device=0,                        # Use GPU
    workers=4,                        # CPU threads
    patience=10,                      # Early stopping
    project='currency_classifier',     # Save folder
    name='yolo_currency',              # Experiment name
    pretrained=True,                   # Use pre-trained weights
    optimizer='AdamW',                  # Optimizer
    lr0=0.001,                            # Initial learning rate
    augment=True,                           # Use augmentation
    verbose=True,                            # Show detailed output
    cos_lr=True,                              # Cosine learning rate
    warmup_epochs=3,                          # Warmup epochs
    label_smoothing=0.0,                       # No label smoothing
    dropout=0.0,                                 # No dropout
)

print("\n✅ Training started! Check the output above for progress.")
```

```python
# Step 4: Monitor Training Progress (Run in a separate cell)
from IPython.display import clear_output
import time
import pandas as pd
import os

log_path = 'currency_classifier/yolo_currency/'
best_acc = 0

print("📊 TRAINING MONITOR - Updates every 30 seconds")
print("Press STOP or Interrupt kernel to stop monitoring\n")

while True:
    try:
        clear_output(wait=True)

        if os.path.exists(f'{log_path}/results.csv'):
            df = pd.read_csv(f'{log_path}/results.csv')

            print("="*60)
            print(f"📈 TRAINING PROGRESS - {time.strftime('%H:%M:%S')}")
            print("="*60)

            # Latest metrics
            latest = df.iloc[-1]
            epoch = int(latest.get('epoch', 0))

            # Calculate progress
            progress = (epoch / 50) * 100
            bar = '█' * int(progress/2) + '░' * (50 - int(progress/2))

            print(f"\nEpoch: {epoch}/50 [{bar}] {progress:.1f}%")
            print(f"\n✅ Top-1 Accuracy: {latest.get('metrics/accuracy_top1', 0):.4f}")
            print(f"✅ Top-5 Accuracy: {latest.get('metrics/accuracy_top5', 0):.4f}")
            print(f"📉 Loss: {latest.get('loss', 0):.4f}")
            print(f"⚡ Learning Rate: {latest.get('lr/lr', 0):.6f}")

            # Track best accuracy
            current_acc = latest.get('metrics/accuracy_top1', 0)
            if current_acc > best_acc:
                best_acc = current_acc
                print(f"\n🏆 NEW BEST ACCURACY: {best_acc:.4f}")

            # Show last 5 epochs
            if len(df) > 5:
                print("\n📊 Last 5 epochs:")
                recent = df[['epoch', 'metrics/accuracy_top1', 'metrics/accuracy_top5', 'loss']].tail()
                print(recent.to_string(index=False))

            # Estimated time remaining (rough estimate)
            if epoch > 0:
                time_per_epoch = 30  # Approximate minutes per epoch
                remaining = (50 - epoch) * time_per_epoch
                hours = remaining // 60
                minutes = remaining % 60
                print(f"\n⏱️  Estimated time remaining: {hours:.0f}h {minutes:.0f}m")

            # Check if best model saved
            if os.path.exists(f'{log_path}/weights/best.pt'):
                size_mb = os.path.getsize(f'{log_path}/weights/best.pt') / (1024*1024)
                print(f"\n💾 Best model saved: {size_mb:.2f} MB")

        else:
            print("⏳ Waiting for training to start...")
            print("Training should begin in a few seconds...")

    except Exception as e:
        print(f"Monitoring... ({str(e)})")

    time.sleep(30)
```
